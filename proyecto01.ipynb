## Generando archivos PDF del presupuesto
* Crear archivo PDF
* Definir el tipo de letra
* Llenar los datos
* Guardar el entregable

```python
!pip install fpdf
```

```python
from fpdf import FPDF
```

```python
pdf = FPDF()

pdf.add_page()
pdf.set_font("Arial")

pdf.text(115, 145, proyecto)
pdf.text(115, 160, horas_estimadas)
pdf.text(115, 175, valor_hora)
pdf.text(115, 190, plazo_estimado)
pdf.text(115, 205, str(valor_total))



pdf.output("presupuesto.pdf")
```

### Inicializar el repositorio Git y subir el proyecto

```python
# 1. Inicializar un nuevo repositorio Git local en la carpeta actual
# Esto crea una subcarpeta .git donde Git almacena todos sus metadatos para el nuevo repositorio.
!git init
```

```python
# 2. Añadir todos los archivos del proyecto al área de staging de Git
# El punto '.' significa añadir todos los archivos y directorios en la carpeta actual.
!git add .
```

```python
# 3. Hacer un commit de los cambios
# Esto guarda los archivos en el historial del repositorio con un mensaje descriptivo.
!git commit -m "Primer commit del proyecto de Python"
```
